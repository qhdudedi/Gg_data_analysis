```python
import pandas as pd

# 1. 쉼표(,) 제거 및 수치형 변환 함수 (문자열로 읽힌 숫자 변환)
def clean_numeric(df):
    for col in df.columns:
        if col != '행정구역':
            df[col] = df[col].astype(str).str.replace(',', '').astype(float)
    return df

df_pop = clean_numeric(df_pop)
df_age = clean_numeric(df_age)
```

```python
# 2. 화도읍 본청 + 동부출장소 + 남부출장소 인구 합산하여 '화도읍'으로 통합
hwado_mask_pop = df_pop['행정구역'].str.contains('화도읍')
hwado_sum_pop = df_pop[hwado_mask_pop].select_dtypes(include=['float', 'int']).sum()
df_pop.loc[df_pop['행정구역'].str.startswith('경기도 남양주시 화도읍('), hwado_sum_pop.index] = hwado_sum_pop.values

hwado_mask_age = df_age['행정구역'].str.contains('화도읍')
hwado_sum_age = df_age[hwado_mask_age].select_dtypes(include=['float', 'int']).sum()
df_age.loc[df_age['행정구역'].str.startswith('경기도 남양주시 화도읍('), hwado_sum_age.index] = hwado_sum_age.values
```

```python
# 16개 동별 5개년 유소년 vs 고령층 증감 계산
df_shift = pd.DataFrame({'행정구역': df_merged['행정구역']})

# 2020년 고령인구 합산
pop_senior_2020 = (df_merged['2020년_계_70~79세'] + df_merged['2020년_계_80~89세'] + 
                   df_merged['2020년_계_90~99세'] + df_merged['2020년_계_100세 이상'])

df_shift['유소년_5개년증감(명)'] = df_merged['2025년_계_0~9세'] - df_merged['2020년_계_0~9세']
df_shift['고령층_5개년증감(명)'] = df_merged['2025_고령인구(70세+)'] - pop_senior_2020

# 고령층 증가율 계산
df_shift['고령층_증가율(%)'] = round((df_shift['고령층_5개년증감(명)'] / pop_senior_2020) * 100, 2)

print("=== [2. 동별 5개년 유소년 vs 고령층 실제 증감표] ===")
display(df_shift.sort_values(by='유소년_5개년증감(명)', ascending=False).reset_index(drop=True))
```

### 2025년 단일 기준 연령대별 인구 분포 확인
- 신도시와 구도심 

```python
# 1. 2025년 기준 3대 생애주기 파생변수 생성
df_merged['2025_유소년(0~9세)'] = df_merged['2025년_계_0~9세']
df_merged['2025_청년신혼(20~39세)'] = df_merged['2025년_계_20~29세'] + df_merged['2025년_계_30~39세']
df_merged['2025_고령층(70세+)'] = (
    df_merged['2025년_계_70~79세'] + df_merged['2025년_계_80~89세'] + 
    df_merged['2025년_계_90~99세'] + df_merged['2025년_계_100세 이상']
)
```

```python
df_merged['2025_유소년비율'] = (df_merged['2025_유소년(0~9세)'] / df_merged['2025년_총인구수']) * 100
df_merged['2025_청년신혼비율'] = (df_merged['2025_청년신혼(20~39세)'] / df_merged['2025년_총인구수']) * 100
df_merged['2025_고령비율'] = (df_merged['2025_고령층(70세+)'] / df_merged['2025년_총인구수']) * 100
```

```python

# 3. 유소년 비율 기준 정렬 (barh에서 위쪽에 큰 값이 오도록 오름차순 정렬 후 렌더링)
df_plot = df_merged.sort_values(by='2025_유소년비율', ascending=True).reset_index(drop=True)

# 3. 누적 가로 막대그래프 시각화
fig, ax = plt.subplots(figsize=(12, 8))

y = df_plot['행정구역']
ax.barh(y, df_plot['2025_유소년비율'], label='유소년(0~9세)', color='#4a90e2', height=0.6)
ax.barh(y, df_plot['2025_청년신혼비율'], left=df_plot['2025_유소년비율'], label='청년/신혼부부(20~39세)', color='#2ecc71', height=0.6)
ax.barh(y, df_plot['2025_고령비율'], left=df_plot['2025_유소년비율'] + df_plot['2025_청년신혼비율'], label='고령층(70세+)', color='#e74c3c', height=0.6)

ax.set_title('2025년 남양주시 연령대별 구성비 비교', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('인구 구성비 (%)', fontsize=11)
ax.legend(loc='upper right', bbox_to_anchor=(1.0, 1.1), ncol=3)
ax.grid(axis='x', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
